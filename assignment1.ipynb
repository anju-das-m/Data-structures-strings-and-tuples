## Strings

### String concatenation

```python
#String concatenation
string1="Hello "
string2=input("Enter your name: ")
output=string1 +string2
print(output)
```

```python
string3=", welcome to Python programming"
string4=output + string3
print(string4)
```

### String slicing and index

```python
# first character
print(string4[0])
```

```python
# Last character
print(string4[-1])
```

```python
#first five character
print(string4[:6])
```

```python
# Last 11 character
print(string4[-11:])
```

```python
# reverse printing
print(string4[::-1])
```

```python
# print the python using slicing
print(string4[23:29])
```

### String Method


```python
strM="Python beginner tutorial"
```

```python
#convert to upper case
print(strM.upper())
```

```python
# convert to lower case
print(strM.lower())
```

```python
# capitalize the sentence
print(strM.capitalize())
```

```python
# count the number of "t" in the string
print(strM.count('t'))
```

```python
# replace 'Python' with 'Machine Learning'
print(strM.replace('Python','Machine Learning'))
```

# Tuple

```python
tuple1=(10,20,30)
tuple2=(40,50,60)
```

```python
# combining the tuple
t_combine=tuple1 + tuple2
print(t_combine)
```

```python
# repeating the elements
print(t_combine * 3)
```

```python
# accessing 3rd element
print(t_combine[2])
```

```python
#access first 3 elements
print(t_combine[:3])
```

```python
# access last 3 elements
print(t_combine[-3:])
```
